```python
#1. Sum of Even and Odd Numbers Get a number n from the user and calculate the sum of all even numbers and the sum of all odd numbers between 1 and n.

n = int(input("Enter a number: "))

even_sum = 0
odd_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

print("Even Sum:", even_sum)
print("Odd Sum:", odd_sum)
```

```python
#2.Count Digits in a Number Get an integer from the user and count the number of digits using a single for loop. Do not convert the number into a string.

n = abs(int(input("Enter a number: ")))
count = 0
if n == 0:
    count = 1
else:
    for i in range(n):
        if n > 0:
            count = count + 1
            n = n // 10

print("Total Digits:", count)
```

```python
#3.Find the Second-Largest Number Given a list of distinct numbers, find the second-largest number without using sort() or max().

numbers = [10, 45, 23, 89, 67]
largest = float('-inf')
second = float('-inf')

for num in numbers:
    if num > largest:
        second = largest
        largest = num

    elif num > second:
        second = num

if second == float('-inf'):
    print("Second largest does not exist")
else:
    print("Second Largest:", second)
```

```python
# 4. Count Vowels and Consonants Get a string from the user and count the vowels and consonants separately. Ignore spaces, digits, and special characters.
text = input("Enter a string: ")
vowels = 0
consonants = 0

for ch in text.lower():
    if ch.isalpha():
        if ch in "aeiou":
            vowels = vowels + 1
        else:
            consonants = consonants + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
# 5.Calculate Factorial Get a positive integer from the user and calculate its factorial using a single for loop. Example: 5 → 120
n = int(input("Enter a number: "))
factorial = 1

if n < 0:
    print("Factorial is not defined")
else:
    for i in range(1, n + 1):
        factorial = factorial * i

    print("Factorial:", factorial)
```

```python
# 7. Calculate Electricity Bill Get the units consumed by each of five customers using a list. Calculate each customer's bill using these rates:
units = []
for i in range(5):
    u = int(input("Enter units: "))
    units.append(u)

for u in units:
    if u <= 100:
        bill = u * 2

    elif u <= 200:
        bill = (100 * 2) + (u - 100) * 3

    else:
        bill = (100 * 2) + (100 * 3) + (u - 200) * 5

    print("Units:", u, "Bill: Rs.", bill)
```

```python
# 8. Reverse a Number Get an integer from the user and reverse its digits using a single for loop. Do not convert it into a string. Example: 12345 → 54321
n = int(input("Enter a number: "))
reverse = 0
num = abs(n)
for i in range(num):
    if num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10

if n < 0:
    reverse = -reverse

print("Reversed Number:", reverse)
```

```python
# 9. Analyze Student Marks Given a list of marks, count the students who passed (50 or above), failed (below 50), and scored distinction (75 or above).
marks = [85, 45, 67, 92, 30, 76, 55]
passed = 0
failed = 0
distinction = 0

for mark in marks:
    if mark >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

    if mark >= 75:
        distinction = distinction + 1

print("Passed:", passed)
print("Failed:", failed)
print("Distinction:", distinction)
```
